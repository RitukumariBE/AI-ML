```python
# Print all the int from 1 to 5.
for i in range(5):
  print(i+1)
```

```python
# using while

i = 1

while i<6:
  print(i)
  i = i + 1
```

```python
# Write a while loop to print HI 7 times on the screen.

i = 1

while i < 8 :
  print("HI")
  i = i + 1
```

```python
i = 1

while i <= 7 :
  print("HI")
  i = i + 1
```

```python
i = 1
while i<=7 :
  print(i,"Hi")
  i += 1 # i = i + 1
```

```python
i = 7
while i>=1 :
  print(i,"Hi")
  i -= 1 # i = i + 1
```

```python
# Print all the evennumbers between 0 to 10 (both inclusive) using a while loop.

i = 0

while i <= 10:
  print(i)
  i = i + 2
```

```python
i = 0

while i <= 10:
  if i % 2 == 0:
    print(i)
  i += 1

```

```python
# Print all the odd numbers between 0 to 10 (both inclusive) using a while loop.

i = 1

while i <= 10:
  print(i)
  i = i + 2
```

```python
i = 0

while i <= 10:
  if i % 2 != 0:
    print(i)
  i += 1
```

```python
# Print the sum of all the numbers between 1 to 10(both inclusive)

i = 1
sum = 0
while i<= 10 :
  sum = sum + i
  i += 1
print(sum)
```

```python
# for loop
sum = 0

for i in range (1,11):
  sum = sum + i
  print(sum)
print(sum)
```

```python
# Print the multiplication table of 5 till its first 10 multiples:

i = 1
n = 5

while i <= 10:
  print(n,"*",i,"=",5*i)
  i += 1

```
