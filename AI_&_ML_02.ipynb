```python
# Take year of birth as input and print the age
bir_year= int(input("Enter the year of birth:"))
curr_year = int(input("Enter the curr_year:"))
age = curr_year - bir_year
print("The current age is :",age)
```

```python
# Take length and breadth of the rectangle as input and print its area.
length = int(input("Enter the length of the rectangle:"))
breadth = int(input("Enter the breadth of the rectangle:"))
area = length * breadth
print("The Area of the rectangle is ",area)
```

```python
# Take length and breadth of the rectangle as input and print its area.
length = float(input("Enter the length of the rectangle:"))
breadth = float(input("Enter the breadth of the rectangle:"))
area = length * breadth
print("The Area of the rectangle is ",area)
```
