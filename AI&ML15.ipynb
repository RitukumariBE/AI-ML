```python
# Given a string , write a code to convert it to lowercase.U cannot use lower().
# A = 65 ,a = 97,Z = 90,z = 122
s = "Raj"
# op : "raj"

new_s = ""
for ch in s:
  if ch <= "Z" and ch >= "A":
    new_s += chr(ord(ch) + 32)
  else:
    new_s += ch

print(new_s)
```

```python
# Given a string , write a code to convert it to uppercase.U cannot use lower().
# A = 65 ,a = 97,Z = 90,z = 122

s = "Raj"
# op : "raj"

new_s = ""
for ch in s:
  if ch <= 'z' and ch >= 'a':
    new_s += chr(ord(ch)-32)
  else:
    new_s += ch

print(new_s)
```
