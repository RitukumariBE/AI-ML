```python
def check(num):
  if num % 2 == 0:
    return "Even"
  else:
    return "Odd"
```

```python
print(check(15))
```

```python
def greet(name):
  print("Hello",name)

greet("Rahul")
```

```python
def total(numbers):
  s = 0

  for n in numbers:
    s += n

  return s

print(total([5,10,15]))
```

```python
def display():
  return "Python"

print(display())
```

```python
def largest(a,b):
  if a > b:
    return a
  else:
    return b

print(largest(12,25))
```
